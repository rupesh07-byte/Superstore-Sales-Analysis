##  Checked data types

```python
print(df.dtypes)
```

** Changed data type of profit_inr

```python
df["profit_inr"] = (
    df["profit_inr"]
    .astype(str)
    .str.replace(",", "", regex=True)
    .str.strip()
)

df["profit_inr"] = pd.to_numeric(df["profit_inr"], errors="coerce")
```

```python
df["sales_inr"] = (
    df["sales_inr"]
    .astype(str)
    .str.replace(",", "", regex=True)
    .str.strip()
)

df["sales_inr"] = pd.to_numeric(df["sales_inr"], errors="coerce")
```

```python
print(df["profit_inr"].dtype)
```

```python
print(df["sales_inr"].dtype)
```

## Sales By Category

```python
import pandas as pd
import matplotlib.pyplot as plt


sales_by_category = df.groupby("category")["sales_inr"].sum()

sales_by_category.plot(
    kind="bar",
    title="Total Sales by Category",
    figsize=(8,5)
)

plt.ylabel("Sales_inr")
plt.show()
```

## Profit by Region

```python
profit_by_region = df.groupby("region")["profit_inr"].sum()

profit_by_region.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Profit Contribution by Region",
    figsize=(6,6)
)

plt.show()
```

## Profit by Sub-category

```python
profit_sub = (
    df.groupby("sub_category")["profit_inr"]
    .sum()
    .sort_values(ascending=False)
)

print(profit_sub)
```

##  Plot

```python
import matplotlib.pyplot as plt

profit_sub.plot(kind="bar", figsize=(10,5))
plt.title("Profit by Sub-Category")
plt.ylabel("Profit (INR)")
plt.xlabel("Sub-Category")
plt.show()
```

## Key Insights
- Technology category contributes the highest sales.
- East region generates the maximum profit.
- Dataset contains ~10,000 retail transactions.
